```python
# Map
li1 = [4,2,6,3,8,7,9]
li2 = list(map(lambda x: x**2, li1))   # map(function, iterable object). Here lambda function has been used
print('\nThe result list = ', li2, '\n')
```

```python

# Filter
li3 = [2,5,3,6,7,3,11,56,73,68]
li4 = list(filter(lambda n: n%2 != 0, li3))   # filter(function, iterable object). Here lambda function has been used
print('\nThe odd list is = ', sorted(li4), '\n')
```

```python
# Reduce
li5 = [2,5,3,6,7,3,1,5,7,8]
from functools import reduce
sm = reduce(lambda a, b: a+b, li5)
print('\nThe sum of the list = ', sm, '\n')
```

```python
from functools import reduce

list1 = [2, 5, 8, 4, 7, 3]
list2 = list(map(lambda x: x**2, list1))
list3 = list(filter(lambda y: y%2==0, list1))
k = reduce(lambda p, q: p+q, list1)
print('The square list is:', list2)
print('The even number is:', list3)
print('The sum is:', k)
```
