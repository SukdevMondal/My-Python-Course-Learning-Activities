```python
print('\n')
x = float(input('Enter a number:'))
if x % 2 == 0:     # Nested if else
    if x % 3 == 0:
        print('The number is dibisible by 6.')
    else:
        print('The number is an even number.')
else:
    print('The number is not an even number.')
```

```python
temp = float(input('Enter the temperature:'))
humd = float(input('Enter the relative humidity:'))
if temp>25:
    if humd>70:
        print('It is hot and humid day.')
    else:
        print('It is hot and non humid day.')
elif temp<25:
    if humd>70:
        print('It is cold and humid day.')
    else:
        print('It is cold and non humid day.')
else:
    print('Invalid.')
```
