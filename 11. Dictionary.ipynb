```python
dev = {
    'Name': 'Sukdev Mondal',
    'Home town': 'Jashore',
    'University': 'BAU',
    'Dept': 'FPM',
}
print('\nMy information is:', dev)
print('The keys are:', dev.keys())
print('The values are:', dev.values(), '\n')
```

```python
dev['Dept'] = 'IWM'      # Change the key value
dev.update({'Home town' : 'Monirampur'})    # Another way to change the key value
dev.pop('University')    # Delete an item from dictionary
del dev['Name']    # Another way to delete the key value
dev['Hall'] = 'BSMH'     # Add a new key
print('The new information is:', dev, '\n')
print('All the keys are:')
```

```python
for x in dev:        # Print the key by a for loop
    print(x)
print('\nAll the values are:')
for y in dev:      # Print the key values by a for loop
    print(dev[x])
    
```

```python

# Dictionary from user inputs
n = int(input('\nEnter the number of elements of your dictionary:'))
d = {}
for i in range (n):
    key = input()
    value = input()
    d[key] = value
print('Your dictionary is:', d)
```
