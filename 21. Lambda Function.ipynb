```python
print('\n')
a = float(input('Enter the first number:'))
b = float(input('Enter the second number:'))
f = lambda a, b: (a**2) + 2*a*b + (b**2)   # lambda arguments: expression
print('\n The result = ', f(a, b), '\n')
```

```python
cu = lambda x: x**3
n = int(input('Enter the number:'))
print('The cuber of', n, 'is:', cu(n))
```
