```python
Data = 'Hello! This is Sukdev Mondal. I am a student of level-4, semester-1. My versity ID is: 2105057 and Reg No is: 53426. Thank you'
print('Data type is: ', type(Data))  # To show the data type
print('Length of data is: ', len(Data), 'letter')  # To show the length of string
print('The ID of the above string is: ', id(Data))   # To find the id of a specific string
import sys
print('Size of data is: ', sys.getsizeof(Data), 'byte')   # To show the size of the data
s = 'semester-1'
print('Semester-1 is located after: ', Data.find(s), 'letter')   # To find a specific string from the main string
print('Is remains about:',Data.count('is'), 'times')   # To show how often a specific string exists into the main string
```

```python
s = 'two B or not 2 b iS THe QuEstiON.'
print('The s string in sentence case is: ', s.capitalize())   # To convert the string in sentense case
print('The s string in swapcase is:', s.swapcase())   # To convert the string in swapcase
print('The s string in title case is:', s.title())   # To capitalize each word in a sentence
print('Are all the characters in uppercase?:', s.isupper())  # To cheque if all the character are in uppercase
print('Are all the characters in lowercase?:', s.islower())  # To cheque if all the character are in lowercase
print('Are all the characters digit?:', s.isdigit())   # To cheque if all the character are in digit
print('All the words of the streing are:', s.split())  # To show all the words of a sentence
print('All the words of the streing are:', s.split() [0])  # To show the first word of a sentence
print('The string is in conter of 50 characters:', s.center(50))  # To show the string in center
print('Replacing the Question by Sukdev:', s.replace('QuEstiON', 'SUKDEV'))  # To replace any word by other
```
