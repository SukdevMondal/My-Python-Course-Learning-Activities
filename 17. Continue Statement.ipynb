```python
print('\nThe odd numbers from 14 to 50 are:')
for i in range (14, 50):
    if i%2 == 0:
        continue
    print(i)
```

```python
print('\n')
dev = 'Om trayambacum yaja mohe , Sugandhim pusti bordhanam , Urbar rukamiba bondhanam , Mritur muksya mamritam.'
ss = dev.split()
for x in ss:
    if x == ',':
        print('\n')
        continue
    print(x)
```
