```python
Name = ['Dev', 'Mohon', 'Badhon', 'Sourob', 'Puja']
Versity = ['BAU', 'DU', 'BU', 'JUST', 'JUST']
Info = list(zip(Name, Versity))     # Making zip from two iterable variable
print('\nThe information of students with their university:\n', Info, '\n')
x, y = zip(*Info)      # Making unzip from a zip file
print('The keys are:', list(x), '\nand \nThe values are:', list(y), '\n')
```

```python
import zipfile as zp
zipp = zp.ZipFile('dev.zip', 'w')     # 'w' means write
zipp.write('111.jpg')
zipp.write('222.jpg')
zipp.write('Introduction.pdf')
```

```python
Name = ('Sukdev', 'Anukul', 'Rana', 'Anika', 'Jitu', 'Shuvo')
Id = (57, 21, 61, 41, 42, 68)
ss = list(zip(Id, Name))
print('\nThe students id with name:', '\n', ss, '\n')

Idn, Namen = zip(*ss)
print('Students id and name is:', '\n', Idn, '\n', Namen, '\n')
```

```python
print('Students name and their id is bellow:')
for Name, Id in ss:
    print(Id, '-', Name)
print('\n')
```

```python
print('Students rank, name and their id is bellow:')
for index, pair in enumerate (ss):
    I, N = pair
    print('Rank- {}, ID- {}, Name: {}'.format(index+1, I, N))
print('\n')
```
