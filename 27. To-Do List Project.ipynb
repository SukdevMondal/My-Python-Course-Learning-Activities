```python
# To Do List
print('Welcome to my interface!')
print('Menue:')
print('1. Add task.')
print('2. View task.')
print('3. Remove task.')
print('4. Exit.')

list1 = []
```

```python
choice = int(input('Enter your choice according to menue:'))

if 4<choice<1:
    print('Invalid choice. Please enter according to menue.')

else:
    if choice == 1:
        task = input('Enter your task:')
        list1.append(task)
        print(f'Your task {task} has been successfully added.')
    
    elif choice == 2:
        if len(list1) == 0:
            print('Your task list is empty.')
        else:
            for i, t in enumerate(list1, start=1):
                print(f"{i}: {t}")

    elif choice == 3:
        if len(list1) == 0:
            print('Your task list is empty.')
        else:
            print('All of your task for remove:')
            for i, t in enumerate(list1, start=1):
                print(f"{i}: {t}")
            n = int(input('Enter the task number to remove:'))
            x = len(list1)
            if x<n<0:
                print('Invalid task number. Please enter according to list.')
            else:
                y = list1.pop(n-1)
                print(f'Your {y} task has been removed successfully.')

    elif choice == 4:
        print('Exited from menue successfully.')
        #break
    
    else:
        print('Invalid Choice. Please enter accoring to the menue.')
```
