```python
x = 57   # Integer number
y = 21.47   # Float number
z = 42.5+35.3j   # Complex number
print('The data type of x, y, and z are: ', '\n', 'For X:', type(x), '\n', 'For Y:', type(y), '\n', 'For Z:', type(z))
```

```python
a = float(x)   # To convert any number into float number
print('The x is converted into float number:', a)
b = int(y)   # To convert any number into integer number
print('The y is converted into integer number:', b)
c = complex(y)  # To convert any number into complex number
print('The y is converted into complex number:', c)
```

```python

l = [2, 5, 6, -5, 'Dev', 'BAU',-6, 3, 'AET',-2, ('Anukul', 9, 4, 'Mondal')]   # List type of data
print('The fourth value of l is:', l[3])
print('The last value of last string of l is: ', l[-1][-1])  # Last value index is -1
print('The value of l between 3 to 8 are:', l[2:8])   # 2:8 means from 3 to 8
print('Is the data type of l list?:', isinstance(l, list))  # To cheque if the data type of a specific variable is list
```
