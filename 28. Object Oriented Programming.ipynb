```python
class dev:
    def __init__ (self, name, id):
        self.name=name
        self.id=id

student1=dev('Sukdev', 2105057)
print(student1.name)
print(student1.id)
```

```python
class dev:
    def __init__(self, name, age, CGPA, faculty):
        self.nm=name    # Public attribute
        self._ag=age    # Protected attribute
        self.__C=CGPA   # Private attribute
        self.fc=faculty
    def show_C(self):
        return self.__C

s1=dev('Sukdev Mondal', 28, 3.667, 'AET')
print('Your name is:', s1.nm)
print('Your age is:', s1._ag)
print('Your department is:', s1.show_C())
```

```python
class animal:
    def sound(self):
        print('Make sound')
class dog(animal):
    def sound(self):
        print('Dog sounds as: Gheu Gheu Gheu.....')
class cat(animal):
    def sound(self):
        print('Cat sounds as: Meu Meu Meu.....')
class cow(animal):
    def sound(self):
        print('Cow sounds as: Hamba Hamba Hamba.....')
class goat(animal):
    def sound(self):
        print('Goat sounds as: Maa Maa Maa.....')
    
```

```python

an1 = cow()
an2 = goat()
an1.sound()
an2.sound()
```
