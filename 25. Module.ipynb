```python
import math
n = int(input('Enter a number:'))
a = math.sqrt(n)
b = math.factorial(n)
print('Root value is:', a)
print('Factorial value is:', b)
```

```python
m = float(input('Enter the base:'))
n = float(input('Enter the power:'))
p = math.pi
r = math.pow(m,n) * p
print('The reselt is:', round(r, 3))
```
