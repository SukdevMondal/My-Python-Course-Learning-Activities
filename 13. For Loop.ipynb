```python
# First method
print('\n')
dev = 'SUKDEV MONDAL'
print('All the letters of dev string are:')
for i in dev:
    print(i)
```

```python
# Second method
print('\n')
l = len(dev)
print('All the letters of dev string are:')
for i in range(l):
    print(dev[i], i)
print('\n')
```

```python
print('\n')
m = [45, 24, 67, 39, 23, 51, 75]
l = len(m)
s1 = 0
for i in range(l):
    s1 = s1 + m[i]
    print(s1)
print('The result in first method is:', s1)
print('\n')
```

```python

m = [45, 24, 67, 39, 23, 51, 75]
s2 = 0
for x in m:
    s2 = s2 + x
    print(s2)
print('The result in second method is:', s2)
print('\n')
```

```python
# Reversing the string elements

print('\n')
data = 'Om! Nama Shivaya.'
n = len(data)
print('There are about', n, 'letter of the string.', '\n')
for x in range(n-1, -1, -1):
    print('The', x+1, 'th value of the string is: ', data[x])
print('\n')
```

```python
print('\n', 'The number from -100 to -10:', '\n')
for i in range(-100, -9, 1):
    print(i)
print('\n')
```

```python
n = int(input('\nEnter a number:'))
f = 1
for i in range (1, n+1):
    f = f*i
print('\nThe factorial of your given number is, n! = ', f, '\n')
```

```python
print('First case of range: ')
x = range(13)  # From 0 to 12
for i in x:
    print(i)

print('\n', '\n', 'Second case of range: ')
y = range(5, 24)  # From 5 to 23
for i in y:    
    print(i)

print('\n', '\n', 'Third case of range: ')
z = range(4, 63, 5)  # From 4 to 62 and increasing by 5
for i in z:    
    print(i)

print('\n', '\n', 'Fourth case of range: ')
a = range(50 , 17 , -2)  # From 50 to 18 and decresing by 3
for i in a:
    print(i)
```
