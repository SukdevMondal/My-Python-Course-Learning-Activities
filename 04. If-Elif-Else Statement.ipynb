```python
print('\n')
m = int(input('Enter your marks:'))
if m>=0 and m<33:
    print('Your grade is: F')
elif m >= 33 and m <= 39:
    print('Your grade is: D')
elif m>= 40 and m<= 49:
    print('Your grade is: C')
elif m>= 50 and m<= 59:
    print('Your grade is: B')
elif m>= 60 and m<= 69:
    print('Your grade is: A-')
elif m>= 70 and m<= 79:
    print('Your grade is: A')
elif m>= 80 and m<= 100:
    print('Your grade is: A+')
else:
    print('Invalid marks')
print('\n')
```
