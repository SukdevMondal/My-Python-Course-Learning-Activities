```python
try:
    print('5'/0)
except ZeroDivisionError:
    print('Zero division error has been occured.')
except NameError:
    print('Name error has been occured.')
except IndexError:
    print('Index error has been occured.')
except TypeError:
    print('Type error has been occured.')
except:
    print('An error has been occured.')
else:
    print('No error has been found.')
```
