```python
n = int(input('\nEnter a number:'))
if n<0:
    sign = -1
else:
    sign = 1
m = abs(n)
rn = 0
while m>0:
    x = m%10
    rn = (rn*10) + x
    m = m//10
rn = rn*sign
print('The reverse number of the given number',n, 'is:', rn, '\n')
```
