```python
#1: Round function
x = float(input('Enter the first number:'))
y = float(input('Enter the second number:'))
z = 3.1416*x*y
s = round(z, 2)   # To show two digit after the point
print('Actual result is:', z)
print('Round value is:', s, '\n')
```

```python
#2: Sort function
n = [23.46, 54, 3.62, 46.01, 24, 5.73]
s = sorted(n)
print('\nSorted number is:', s, '\n')
```

```python
#3: Join function
l = ['45', '23', '7', '34', '62', '11']
j = "+".join(l)
print('\nDo the sum:\n', j)
```

```python
#4: Upper and lower function
dev = 'Om trayambacum yaja mohe , Sugandhim pusti bordhanam , Urbar rukamiba bondhanan , Mritur muksya mamritam.'
u = dev.upper()
l = dev.lower()
print('\n In upper case:\n', u , '\n\n','In lower case:\n', l , '\n')
```

```python
#5: Enumerate function
Name = ['Dev', 'Anukul', 'Tarok', 'Akhi', 'Rakhi', 'Shubro', 'Alok']
Roll = ['43', '54', '65', '12', '4', '6', '13']
print('Result of half yearly exam class 7:\n')
z = list(zip(Name, Roll))
for index, pair in enumerate (z):
    N, R = pair
    print('Rank-{}, Name:{}, Roll-{}'.format(index+1, N, R))
```

```python
#6: Sum function
n = [3,6,7,2]
s = sum(n, 1000)   # Adding extra 1000 with sum
print('My current balance is:', s)
```

```python
#7: Slice function
l = ['45', '23', '7', '34', '62', '11']
m = slice(2, 5)   # Taking from 2 to 4 index
s = l[m]
print('My slice is:', s)
```

```python

#8: chr and ord functions
print('The ASCI value of G is:', ord('G'))
print('The chr of 9158 ASCI value is:', chr(9158))
```
