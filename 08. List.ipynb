```python
# Without condition
a = [x*3 for x in range(5,17)]
print('\nMy a list is:', a)

# With condition
b = [y**2 for y in range(7, 23) if y % 2==0]
print ('\nMy b list is:', b)

# With a string
c = 'Om! Namah Sivaya.'
d = [z for z in c if z != 'm']
print('The elements except m are:', d)

# Nested condition
e = [[j**3 for j in range(6, 11)] for i in range(5)]
print('The e list is:\n', e)

# Split function
f = c.split()
print('The f list is', f)

# Condition in a string
g = [len(i) for i in f]
print('The length of all word in e is:', g)
```

```python

l1 = ['dev', 'om', 45, 119, (43, 'tum', 'may'), 'bau', 44]
# Copy function
print('\n')
l2 = l1.copy()    
print('The new list is:', l2)
```

```python
# Replace function
l2[2] = 'anu'
print('The new list is:', l2)
# Append function
l2.append(['AET', 4, 366,'dev'])
print('The new l2 is:', l2)
```

```python
# Extend function
l3 = l1.copy()
l3.extend(['AET', 4, 366,'dev'])
print('The new l3 is:', l3)
# Insert function
```

```python
l4 = l1.copy()
l4.insert(2, 'FPM')
print('The new l4 is:', l4)
```

```python
# Replace function
l5 = ['dev', 45 , 55, 'u9', 'dev']
l5[2] = 'iwm'
print('The new l5 is:', l5)
```

```python
# Index function
x = l5.index(45)
print('The index number of 45 is:', x)
# Count function
y = l5.count('dev')
print('The dev element exists in:', y, 'times')
# Remove function (Erasing an element)
l5.remove('u9')
print('The new l5 is:',l5)
```

```python
# Sort function (Sorting in ascending order)
l6 = [45, 0, 2, -5, 64, 77]
l6.sort()
print('The l6 is sorted in ascending ordr:', l6)
# Reverse function (Sorting i decending order)
l6.reverse()
print('The l6 is sorted in descending order:', l6)
# Pop function (Removing an element by index)
l6.pop(3)
print('The new l6 is:', l6)
```

```python
# Del keyword
l7 = ['DEV', 43, -47, 23, 'BAU', 35, 87]
del l7[2]
print('The new l7 is:', l7)
# Clear function
l7.clear()
print('The new l7 is:', l7)
print('\n')
```
