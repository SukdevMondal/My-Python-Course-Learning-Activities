```python
import array as ar   # It is a library and I defined it as ar for my work
a1 = ar.array('i', [2, 4, 5, 9, 74, 20, 88, 14, 47, 57])   # Here f is the type code of float number
print('\nThe elements of a1 from 2 to 5 positions:', a1[2:5])
a1.append(33)   # Adding a value in last position
a1.pop(7)   # Removing a value by it's index number
a1.remove(47)   # Removing a value by mentioning
a1.reverse()  # Reversing the sequence of elements
print('\nThe a1 array is:', a1)
```
