```python
# Reading text from a file
print('\n')
with open('sss.txt', 'r') as file:   # Read  each line in the file, 'r' means read
    for i in file:
        print(i.strip())    # The strip() function removes any leading or trailing whitespace and newline character
print('\n')
```

```python

# Appending text to an existing file
s = [' ', 'Om! Namah Sivaya.']
with open('Write file-52.txt', 'a') as file:   # 'a' means append
    for y in s:
        file.write(y + '\n')
```

```python
# Reading any page from a pdf file
import PyPDF2   # PyPDF2 is a library
sss = 'D:\Academic\L-4, S-2   (Current)\Courses\Environmental Impact\EIA Lecture Notes.pdf'   # File path
n = int(input('\nEnter the page number you want ot read : '))
with open(sss, 'rb') as file:
    reader = PyPDF2.PdfReader(file)   # Reader function
    page = reader.pages[n-1]     # It reads page (n+1) if input n
    text = page.extract_text()   # Keeping the text in this variable
    print('\n', text, '\n')      # Printing the text
```

```python

# Reading pages by loop from a pdf file
import PyPDF2   # PyPDF2 is a library
sss = "D:\Academic\L-4, S-1\Courses\Agricultural Mechanization and Engineering Management\Theory\zzz---Agril Mechanization Def & Myths-1.pdf"   # File path
with open(sss, 'rb') as file:
    reader = PyPDF2.PdfReader(file)   # Reader function
    for s in reader.pages:     # Here 's' is a variable
        text = s.extract_text()   # Keeping the text in this variable
        print('\n*** Next Page ***\n')
        print('\n', text, '\n')     
```
