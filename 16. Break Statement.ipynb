```python
print('\nThe multiplication table of 9:\n')
for i in range(1, 11):
    if i==11:
        break
    print('9*', i, '=', i*9)
print('\n')
```

```python
print('The prime number from 7 to 50 are:')
for x in range (7, 50):
    if x>1:
        for y in range (2, int(x**0.5)+1):
            if x % y == 0:
                break
        else:
            print(x)
```
