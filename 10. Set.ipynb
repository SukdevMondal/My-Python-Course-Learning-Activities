```python
# Set can not contain duplicate value

set1 = {2, 34, 75, 'sms', False, 'ai', 657}
set1.update(['anu', 92])   # To add multiple values in any set
set1.add('Sivaya')   # To add single value in any set
print('\nThe set1 is:', set1)
set2 = {34, 'sms', 59, 'Om'}
set2.discard(59)   # To delete element from a set
print('The new set2 is:', set2, '\n')
```

```python
s3 = {2, 34, 75, 'sms', False, 'ai', 57}
s4 = {34, 'sms', 57, 'Om'}
s5 = s3.union(s4)   # To do union two set
s6 = s3.intersection(s4)   # To do intersection two set
print('\nThe s3 union s4 is:', s5, '\nThe s3 intersection s4 is:', s6, '\n')
```

```python
# Frozen set do not support update, add, modify, discard or remove any element

l1 = (54, 64, 88, 'FPM', 'BAU', 'DEV', True, 563)
s2 = frozenset(l1)   # l1 list is converted as forzenset
print('\nThe l1 list is converted into:', s2, '\nAnd its type is:', type(s2), '\n')
```

```python
dev = 'Om! Namah Sivaya.'
s3 = [tuple(x) for x in dev]
s4 = [set(y) for y in dev]
s5 = [list(z) for z in dev]
print('The string in tuple:', s3, '\n')
print('The string in set:', s4, '\n')
print('The string in list:', s5, '\n')
```

```python
s1 = {12, 63, 'dev', True, 'fao', 'BAU', 569, 'anu'}
s2 = { 'dev', False, 'anu', True, 32, 63, 'FPM', 'BAU'}
s3 = s1.difference(s2)   # Determine the exception of s1 from s2
s4 = s2.difference(s1)    # Determine the exception of s2 from s1
s5 = s1.symmetric_difference(s2)   # Determine the set by neglecting the common elements
print('\nThe exception of s1 from s2 is:', s3)
print('\nThe exception of s2 from s1 is:', s4)
print('\nThe symmetric differrence of s1 and s2 is:', s5, '\n')
```
