```python
print('\n')
def fact():
    r = 1
    a = int(input('Enter the number: '))
    if a == 0 or a == 1:
        print('The result = 1\n')
    else:
        for i in range (a):
            r = r * (i+1)
        print('The result is = ', r, '\n')
fact()
```

```python
def ss(il):
    ul = []
    for x in il:
        if x not in ul:
            ul.append(x)
    print('\n Your unit list is:', sorted(ul), '\n')
    return ul

il = [3, 4, 2, 6, 3, 4, 2, 7, 2, 9] 
ss(il)
```

```python
# Assignment-1: Iput and sum of a list
def ss():
    li1 = list(map(int, input('\nEnter the members of the list:').split()))   # Input the list
    s = 0
    for i in li1:
        s = s + i
    print('\nYour list is:', sorted(li1))
    print('Sum of the list = ', s, '\n')   
ss()
```

```python
# Assignment-2: Input and reverse a list
def rev_li():
    li2 = list(map(int, input('\nEnter the members of the list:').split()))   # Input the list
    li2.reverse()
    print('Your reversed list is:',li2, '\n')
    return li2
rev_li()
```

```python

# Assignment-3: Input a list and making a new list with even elements 
def ev_li():
    li3 = []
    li4 = list(map(int, input('\nEnter the members of the list:').split()))   # Input the list
    for i in li4:
        if i % 2 == 0:
            li3.append(i)
    print('The even list is:', li3, '\n')
ev_li()
```

```python

# Assignment-4: Input a list and making a new list with unit elements
def uli():
    li5 = []
    li6 = list(map(int, input('\nEnter the members of the list:').split()))
    for i in li6:
        if i not in li5:
            li5.append(i)
    print('The unit list is:', sorted(li5), '\n')
uli()
```

```python
# Assignment-5: Input two list and combining into a single list
def com_li():
    li7 = list(map(int, input('\nEnter the members of the first list:').split()))
    li8 = list(map(int, input('\nEnter the members of the second list:').split()))
    li9 = li7 + li8
    print('\nThe first list is:', li7)
    print('\nThe second list is:', li8)
    print('\nThe combined list is:', li9, '\n')
com_li()
```

```python
def factorial(n):
    if n==1:
        return 1
    return n*factorial(n-1)

x = int(input('Enter the number:'))
print('The factorial of', x, 'is:', factorial(x) )
```
