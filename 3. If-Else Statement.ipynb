```python
print('\n')
t = int(input('Enter the total number of classes:'))
a = int(input('Enter the number of classes you attended:'))
p= float((a/t)*100)
print('Your attancance is:', p , '%')
if p<75:
    print('Sorry, you will not be allowed to sit in the exam.')
else:
    print('Congratulation, you will be allowed to sit in the exam.')
print('\n')
```
