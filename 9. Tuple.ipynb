```python
# No elemen  canbe added to or delet from a tuple directly. For this reason the tuple should be converted into a list.
t1 = ('dev', True, 'anu', 57, 3456, False, (34, 'bau', 43), 'fpm', [11, False, 'Sivaya'])
l1 = list(t1)   # t1 tuple is converted into list
print('\nThe l1 list is:', l1)
l1.insert(1, 'AET')   # Adding a value into the list at 2 position
t2 = tuple(l1)   # l2 list is again converted into tuple
print('The t1 tuple is:', t1)
print('The new l1 list is:', l1)
print('The t2 tuple is:', t2)
```
